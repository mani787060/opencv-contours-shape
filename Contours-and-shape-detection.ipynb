```python
import cv2
```

```python
# for reading an image 
img = cv2.imread("triangle image.png")

# converting image to 'gray-scale'
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# make it a binary_image (i.e, black and white image)
_, thresh = cv2.threshold(gray, 200, 255, cv2.THRESH_BINARY)

# finding contours
contours, hierarchy = cv2.findContours(thresh, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)

# drawing contours
cv2.drawContours(img, contours, -1, (0,0,255), 3)

# printing an image
cv2.imshow("Contours", img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# detecting the shape of an image :-

# for reading an image 
img = cv2.imread("square image.jpg")

# converting image to 'gray-scale'
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# make it a binary_image (i.e, black and white image)
_, thresh = cv2.threshold(gray, 200, 255, cv2.THRESH_BINARY)

# finding contours
contours, hierarchy = cv2.findContours(thresh, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)

# drawing contours
#cv2.drawContours(img, contours, -1, (0,0,255), 3)


for contour in contours:
    approx = cv2.approxPolyDP(contour, 0.01 * cv2.arcLength(contour, True), True)

    # for taking the no. of sides we use 'length' function
    corners = len(approx)       # it gives the no. of corners present in the shape

    if corners == 3:
        shape_name = "Triangle"

    elif corners == 4:
        shape_name = "Rectangle"
        
    elif corners == 5:
        shape_name = "Pentagon"
        
    elif corners > 5:
        shape_name = "Circle"

    else:
        shape_name = "unknown"


    cv2.drawContours(img, [approx], 0, (0,255,0), 2)    

    # for working on the 1st shape
    x = approx.ravel()[0]          # ravel() -> converts multi-D array into 1-D array.
    y = approx.ravel()[1] - 10

    # printing the 'text_name' a/c to it's shape :-
    cv2.putText(img, shape_name, (x,y), cv2.FONT_HERSHEY_COMPLEX, 0.6, (255,0,0), 2)
    

# printing an image
cv2.imshow("Contours", img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2

# for reading an image 
img = cv2.imread("circle image.jpg")

# converting image to 'gray-scale'
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# make it a binary_image (i.e, black and white image)
# If your background is white, 200 is good. 
_, thresh = cv2.threshold(gray, 200, 255, cv2.THRESH_BINARY_INV) 

# FIX: Changed RETR_TREE to RETR_EXTERNAL to get only 1 contour per shape
contours, hierarchy = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

for contour in contours:
    # Filter out tiny noise (dots) so they aren't labeled as "unknown"
    if cv2.contourArea(contour) < 500:
        continue

    # Approximate the shape
    approx = cv2.approxPolyDP(contour, 0.01 * cv2.arcLength(contour, True), True)

    # Number of corners
    corners = len(approx)

    if corners == 3:
        shape_name = "Triangle"
    elif corners == 4:
        # Extra check: Is it a perfect square or a rectangle?
        x, y, w, h = cv2.boundingRect(approx)
        aspectRatio = float(w)/h
        if 0.95 <= aspectRatio <= 1.05:
            shape_name = "Square"
        else:
            shape_name = "Rectangle"
    elif corners == 5:
        shape_name = "Pentagon"
    elif corners > 5:
        shape_name = "Circle"
    else:
        shape_name = "Unknown"

    # Drawing the approximated shape
    cv2.drawContours(img, [approx], 0, (0, 255, 0), 2)  

    # Getting coordinates for text
    x = approx.ravel()[0]
    y = approx.ravel()[1] - 10

    # Printing the name
    cv2.putText(img, shape_name, (x, y), cv2.FONT_HERSHEY_COMPLEX, 0.6, (255, 0, 0), 2)

# Show the final result
cv2.imshow("Contours", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2

# 1. Load the image
img = cv2.imread("square image.jpg")

# 2. Convert to grayscale and threshold
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Using THRESH_BINARY_INV works best for shapes on light backgrounds
_, thresh = cv2.threshold(gray, 200, 255, cv2.THRESH_BINARY_INV)

# 3. Find contours using RETR_EXTERNAL (Fixes the "Double Name" issue)
contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

for contour in contours:
    # Ignore tiny noise/specs
    if cv2.contourArea(contour) < 500:
        continue

    # Approximate the shape
    approx = cv2.approxPolyDP(contour, 0.01 * cv2.arcLength(contour, True), True)
    corners = len(approx)

    # Determine shape name
    if corners == 3:
        shape_name = "Triangle"
    elif corners == 4:
        shape_name = "Rectangle"
    elif corners == 5:
        shape_name = "Pentagon"
    elif corners > 5:
        shape_name = "Circle"
    else:
        shape_name = "Unknown"

    # Draw the contour on the original image
    cv2.drawContours(img, [approx], 0, (0, 255, 0), 2)

    # Get coordinates for the text
    x = approx.ravel()[0]
    y = approx.ravel()[1] - 10
    cv2.putText(img, shape_name, (x, y), cv2.FONT_HERSHEY_COMPLEX, 0.6, (255, 0, 0), 2)



#      --- THE FIX FOR THE SMALL WINDOW ---



# Option A: Resize the image physically (Scale by 3x)
img_big = cv2.resize(img, (0, 0), fx=3, fy=3) 

# Option B: Allow manual resizing of the window
cv2.namedWindow("Contours", cv2.WINDOW_NORMAL) 

# Display the big version
cv2.imshow("Contours", img_big)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
